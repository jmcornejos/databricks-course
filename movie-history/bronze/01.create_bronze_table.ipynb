```python
%sql
CREATE TABLE IF NOT EXISTS moviehistory.bronze.movies (
  movieId INT,
  title STRING,
  budget DOUBLE,
  homePage STRING,   
  overview STRING,
  popularity DOUBLE,
  yearReleaseDate INT,
  releaseDate DATE,
  revenue DOUBLE,
  durationTime INT,
  movieStatus STRING,
  tagline STRING,
  voteAverage DOUBLE,
  voteCount INT
)
USING CSV
OPTIONS(path '/Volumes/moviehistory/bronze/vol_bronze/movie.csv', header true) 
```

```python
%sql
CREATE TABLE IF NOT EXISTS moviehistory.bronze.movies
USING CSV
OPTIONS (
  path "/Volumes/moviehistory/bronze/vol_bronze/movie.csv",
  header "true",
  inferSchema "true",
  delimiter ","
);
```
